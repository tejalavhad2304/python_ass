```python
#9. Using the following array, find the maximum value, minimum value, index of maximum value, and index of minimum value. [45, 12, 89, 34, 67, 23, 91, 56, 8, 72] 
import numpy as np 
class Find_Value:
    def max_min():
        arr= np.array([45, 12, 89, 34, 67, 23, 91, 56, 8, 72] )
        print(np.max(arr))
        print(np.min(arr))
        print(np.argmax(arr))
        print(np.argmin(arr))
obj= Find_Value
obj.max_min()
```
