## Project: Student Performance Analysis 

### Objective :- Create a Python program that analyzes students' marks using Python and NumPy. 

#### Dataset 

```python
import numpy as np

students = np.array([
    ["Umesh", 85, 90, 78],
    ["Rakesh", 72, 65, 80],
    ["Amit", 90, 88, 95],
    ["Priya", 55, 60, 58],
    ["Neha", 92, 95, 90],
    ["Rahul", 40, 35, 45],
    ["Sneha", 76, 82, 79],
    ["Kiran", 65, 70, 68]
])
```

```python
# Q1. Display the complete dataset
print(students)
```

```python
# Q2. Display only student names
print(students[:,0])
```

```python
#3. Display Python marks.
print(students[:,1])
```

```python
#4. Display SQL marks. 
print(students[:,2])
```

```python
#5. Display NumPy marks.
print(students[:,3])
```

```python
#6. Find the total number of students. 
print(len(students))
```

```python
#7. Find the total marks obtained by each student. 
for i in students:
    Total=int(i[1])+int(i[2])+int(i[3])
    print(i[0],"-",Total)
```

```python
#8. Find the average marks of each student.
for i in students:
    Total=int(i[1])+int(i[2])+int(i[3])
    average =Total/3
    print(i[0],"-",average)
```

```python
#9. Find the highest Python mark.
python_marks=students[:,1].astype(int)
highest_python=np.max(python_marks)
print("python highest marks","=",highest_python)
```

```python
#10. Find the lowest Python mark. 
python_marks=students[:,1].astype(int)
lowest_python=np.min(python_marks)
print("python lowest marks","=",lowest_python)
```

```python
#11.Find the highest SQL mark. 
sql_marks=students[:,2].astype(int)
highest_sql=np.max(sql_marks)
print("sql highest marks","=",highest_sql)
```

```python
#12. Find the lowest SQL mark. 
sql_marks=students[:,2].astype(int)
lowest_sql=np.min(sql_marks)
print("sql lowest marks","=",lowest_sql)
```

```python
#13. Find the highest NumPy mark
numpy_marks=students[:,3].astype(int)
highest_numpy=np.max(numpy_marks)
print("numpy highest marks","=",highest_numpy)
```

```python
#14. Find the lowest NumPy mark
numpy_marks=students[:,3].astype(int)
lowest_numpy=np.min(numpy_marks)
print("numpy lowest marks","=",lowest_numpy)
```

```python
#15. Find the average marks of the Python subject. 
python_marks=students[:,1].astype(int)
average_python=np.mean(python_marks)
print("python average marks =",average_python)
      
```

```python
#16. Find the average marks of the SQL subject.
sql_marks=students[:,2].astype(int)
average_sql=np.mean(sql_marks)
print("sql average marks =",average_sql)
```

```python
#17. Find the average marks of the numpy subject.
numpy_marks=students[:,3].astype(int)
average_numpy=np.mean(numpy_marks)
print("numpy average marks =",average_numpy)
```

```python
#18. Find the student who scored the highest Python marks. 
python_marks=students[:,1].astype(int)
highest_python=np.max(python_marks)
for i in students:
    if int(i[1])==highest_python:
        print("who scored the highest Python marks","=",i[0])
```

```python
#19.Find the student who scored the lowest Python marks.
python_marks=students[:,1].astype(int)
lowest_python=np.min(python_marks)
for i in students:
    if int(i[1])==lowest_python:
        print("who scored the lowest Python marks","=",i[0])
```

```python
#20. Find the student with the highest overall average. 
marks = students[:, 1:].astype(int)

averages = np.mean(marks, axis=1)

highest_average = np.max(averages)

index = np.argmax(averages)

print("Student with highest overall average =", students[index, 0])
print("Highest average =", highest_average)
```
