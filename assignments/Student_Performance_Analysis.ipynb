## Project: Student Performance Analysis 

### Objective :- Create a Python program that analyzes students' marks using Python and NumPy. 

#### Dataset 

```python
import numpy as np

students = np.array([
    ["Umesh", 85, 90, 78],
    ["Rakesh", 72, 65, 80],
    ["Amit", 90, 88, 95],
    ["Priya", 55, 60, 58],
    ["Neha", 92, 95, 90],
    ["Rahul", 40, 35, 45],
    ["Sneha", 76, 82, 79],
    ["Kiran", 65, 70, 68]
])
```

```python
#7. Find the total marks obtained by each student. 
for i in students:
    Total=int(i[1])+int(i[2])+int(i[3])
    print(i[0],"-",Total)
```

```python
#8. Find the average marks of each student.
for i in students:
    Total=int(i[1])+int(i[2])+int(i[3])
    average =Total/len(i[1:])
    print(i[0],"-",average)
```

```python

#18. Find the student who scored the highest Python marks. 
python_marks=students[:,1].astype(int)
highest_python=np.max(python_marks)
for i in students:
    if int(i[1])==highest_python:
        print("who scored the highest Python marks","=",i[0])
```

```python
#19.Find the student who scored the lowest Python marks.
python_marks=students[:,1].astype(int)
lowest_python=np.min(python_marks)
for i in students:
    if int(i[1])==lowest_python:
        print("who scored the lowest Python marks","=",i[0])
```

```python
#20. Find the student with the highest overall average. 
marks = students[:, 1:].astype(int)

averages = np.mean(marks, axis=1)

highest_average = np.max(averages)

index = np.argmax(averages)

print("Student with highest overall average =", students[index, 0])
print("Highest average =", highest_average)
```

```python
# Q.20 using python
highest_average = 0
highest_student = ""

for i in students:
    total = int(i[1]) + int(i[2]) + int(i[3])
    average = total / 3

    if average > highest_average:
        highest_average = average
        highest_student = i[0]

print("Student with highest overall average =", highest_student)
print("Highest average =", highest_average)
```

```python
#20. Find the student with the lowest overall average. 
marks = students[:, 1:].astype(int)

averages = np.mean(marks, axis=1)

lowest_average = np.min(averages)

index = np.argmin(averages)

print("Student with lowest overall average =", students[index, 0])
print("lowest average =", lowest_average)
```

```python
# Q.20 using python
lowest_average = 100
lowest_student = ""

for i in students:
    total = int(i[1]) + int(i[2]) + int(i[3])
    average = total / 3

    if average <lowest_average:
        lowest_average = average
        lowest_student = i[0]

print("Student with lowest overall average =", lowest_student)
print("lowest average =", lowest_average)
```
