```python
class Numbers:
    def print_1_to_10(self):
        for i in range(1,11):
            print(i)
obj=Numbers()
obj.print_1_to_10()
            
```

```python
class ReverseNumbers:
    def print_reverse(self):
        for i in range(10,0,-1):
            print(i)
obj=ReverseNumbers()
obj.print_reverse()
```

```python
class EvenNumber:
    def print_even(self):
        for i in range(1,100):
            if i % 2 == 0:
                print(i,end=" ")


obj = EvenNumber()
obj.print_even()
```

```python
class OddNumber:
    def print_odd(self):
        for i in range(1,100):
            if i % 2 != 0:
                print(i,end=" ")


obj = OddNumber()
obj.print_odd()
```

```python
class Table:
    def multiplication_table(self,n):
        for i in range(1,11):
            print(n,"x",i,"=", n*i)
n = int(input("Enter your number :"))
obj = Table()
obj.multiplication_table(n)
```

```python
class NaturalSum:
    def calculate_sum(self, n):
        total = 0

        for i in range(1, n + 1):
            total = total + i

        print("Sum =", total)


n = int(input("Enter n: "))
obj = NaturalSum()
obj.calculate_sum(n)
```

```python
# 7. Sum of all even numbers from 1 up to n

class EvenSum:
    def calculate_sum(self, n):
        total = 0

        for i in range(1, n + 1):
            if i % 2 == 0:
                total = total + i

        print("Even Sum =", total)


n = int(input("Enter n: "))
obj = EvenSum()
obj.calculate_sum(n)
```

```python
# 8. Sum of all odd numbers from 1 up to n

class OddSum:
    def calculate_sum(self, n):
        total = 0

        for i in range(1, n + 1):
            if i % 2 != 0:
                total = total + i

        print("Odd Sum =", total)


n = int(input("Enter n: "))
obj = OddSum()
obj.calculate_sum(n)
```

```python
# 9. Factorial of a given number

class Factorial:
    def calculate_factorial(self, n):
        fact = 1

        for i in range(1, n + 1):
            fact = fact * i

        print("Factorial =", fact)


n = int(input("Enter number: "))
obj = Factorial()
obj.calculate_factorial(n)
```

```python
# 10. Product of all digits

class DigitProduct:
    def calculate_product(self, n):
        product = 1

        while n > 0:
            digit = n % 10
            product = product * digit
            n = n // 10

        print("Product of digits =", product)


n = int(input("Enter number: "))
obj = DigitProduct()
obj.calculate_product(n)
```
